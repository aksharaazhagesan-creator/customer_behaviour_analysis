```python
import pandas as pd
```

```python
df = pd.read_csv("../customer_shopping_behavior.csv")
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
# divides the dataset into groups based on categories and for each category only see review rating column and fill up the missing values with the median
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
#create a column age_group to group customers into 4 age groups,q divides ages into 4 equal groups and assigns the labels we defined 
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels = labels)
                          #the qcut depends on our dataset ...it divides all the ages into 4 equal groups
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchse_frequency_days
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monhtly' : 30,
    'Quaterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
} #we convert textual frequency to number of days
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
#using map function we shall replace the text to numbers in tht column
```
